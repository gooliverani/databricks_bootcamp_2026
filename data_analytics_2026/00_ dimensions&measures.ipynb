#Dimensions & Measures

####Retrieve a list of unique categories of products
- ##### Dimension (Make no sense to aggregate)

```python
query = """
SELECT DISTINCT
    category
FROM gold.dim_products
"""
df = spark.sql(query)

df.display()
```

####Retrieve a list of unique sales amount
- ##### Measure (Make sense to aggregate)

```python
query = """
SELECT DISTINCT
    sales_amount
FROM gold.fact_sales
"""
df = spark.sql(query)

df.limit(10).display()
```

####Retrieve a list of unique product names
- ##### Dimension (Make no sense to aggregate)

```python
query = """
SELECT DISTINCT
    product_name
FROM gold.dim_products
"""
df = spark.sql(query)

df.limit(10).display()
```

#### Retrieve a list of unique quantity of sales
- ##### Measure (Make sense to aggregate)

```python
query = """
SELECT DISTINCT
    quantity
FROM gold.fact_sales
"""
df = spark.sql(query)

df.display()
```

####'birthdate' is Dimension (Make no sense to aggregate)
####But when you calculate the age from the 'birthdate' it's Measure
- ##### Measure (Make sense to aggregate)

```python
query = """
SELECT DISTINCT
    ROUND(AVG(DATEDIFF(year, birthdate, GETDATE())), 0) AS Avg_Age
FROM gold.dim_customers
"""
df = spark.sql(query)

df.display()
```
